# Initial EDA on supply chain data 

1. Download the data from here: [data source from kaggle](https://www.kaggle.com/datasets/shashwatwork/dataco-smart-supply-chain-for-big-data-analysis)

2. Create a catalog, schema and volume inside of Unity Catalogs and upload the csv files into the volume 

3. Read it in and start EDA


I'll make a very short EDA here, just so that you can get started, you should definitely do a more complete EDA on this dataset

## Setup catalog and schema 

```python
%sql 

DROP TABLE IF EXISTS supply_chain.bronze.raw_supply_chain;

CREATE CATALOG IF NOT EXISTS supply_chain;

CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
```
